```python
numbers = [1, 2, 3, 4, 5, -6, -7, 8]

counter = 0

for num in numbers:
    if num > 0:
        counter += num        

print(counter)
```

```python
slova = ["Python", "algoritmus01", "proměnná", "funkce", "struktura", "algoritmus02"]

longest_word = ""

for word in slova:
    if len(word) > len(longest_word):
        longest_word = word

print(longest_word)
print(len(longest_word))
    
```

```python
import random

numbers = []

for num in range(25):
    random_number = random.uniform(5, 25)
    random_number = round(random_number, 1)
    numbers.append(random_number)

print(numbers)
```

```python
smallest_number = 50
for num in numbers:
    if num < smallest_number:
        smallest_number = num

print(smallest_number)
```
